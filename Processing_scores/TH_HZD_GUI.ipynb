# **RDL - Multi-Hazard Threshold Analysis**

## Multi-hazard threshold-based scoring for administrative units

<p>Produces hazard scores [0-3] for administrative units based on value and area thresholds for multiple natural hazards:</p>

### Supported Hazards:
- **Earthquake**: GAR17 PGA data (RP 250, RP475, RP975, RP2475)
- **Landslide**: Index-based classification (1-5 scale)
- **Tsunami**: Work in progress
- **Volcano**: VEI (Volcanic Explosivity Index) buffered zones
- **Cyclone**: STORM v4 wind speed (RP10, RP100, RP1000, RP10000)
- **Extreme Heat**: WBGT temperature (RP5, RP20, RP100)
- **Wildfire**: FWI (Fire Weather Index) (RP5, RP25, RP50)
- **Water Scarcity**: Work in progress

### Boundary Options:
- **ADM2**: Global administrative level 2 boundaries (via WorldBank REST API)
- **Urban**: Local urban boundaries (data/ADM/TH_urban.gpkg)
